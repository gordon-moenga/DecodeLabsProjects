This project focuses on analyzing an e-commerce order dataset to uncover patterns in customer purchases, products, payments, and order activity.
```python
# converting date column dtype into datetime
dataset['Date'] = pd.to_datetime(dataset['Date'])
```

```python
# Replace all missing values in the coupon column with 'No Coupon'
dataset['CouponCode'] = dataset['CouponCode'].fillna('No Coupon')
```

```python
# check whether total price is consistent with unit priice and quantity
calculated_total = dataset['Quantity'] * dataset['UnitPrice']

(dataset['TotalPrice'] == calculated_total).sum()
```

```python
dataset['CalculatedTotal'] = dataset['Quantity'] * dataset['UnitPrice']

dataset['PriceDifference'] = (
    dataset['TotalPrice'] - dataset['CalculatedTotal']
)

dataset[dataset['PriceDifference'] != 0][
    ['OrderID', 'Quantity', 'UnitPrice', 'TotalPrice',
     'CalculatedTotal', 'PriceDifference']
].head(10)
```

```python
calculated_total = dataset['Quantity'] * dataset['UnitPrice']

matches = np.isclose(
    dataset['TotalPrice'],
    calculated_total
)

matches.sum()
```
